```python
# 轉成 mp3（小很多，網頁直接內嵌）並打包
import glob, os, subprocess, zipfile
os.makedirs('/kaggle/working/mp3', exist_ok=True)
for w in sorted(glob.glob('/kaggle/working/wav/*.wav')):
    out = '/kaggle/working/mp3/' + os.path.basename(w)[:-4] + '.mp3'
    subprocess.run(['ffmpeg', '-y', '-loglevel', 'error', '-i', w, '-ac', '1', '-b:a', '48k', out], check=True)
zp = '/kaggle/working/11_語音.zip'
with zipfile.ZipFile(zp, 'w') as z:
    for m in sorted(glob.glob('/kaggle/working/mp3/*.mp3')):
        z.write(m, os.path.basename(m))
print('完成：', len(glob.glob('/kaggle/working/mp3/*.mp3')), '/', 21, '個檔 →', zp)
```
